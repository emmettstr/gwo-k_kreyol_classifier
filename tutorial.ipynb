# Using Gwo-K Kreyol Classifier

### The model can be imported using the joblib library

```python
import joblib
model = joblib.load("models/gwo_k.joblib")
```

### For inference, the model can be applied to a list of texts

```python
texts = [
    "Gwo-K is a lightweight n-gram based classifier for identifying French-lexifier creoles.",
    "Ce modèle peut également distinguer ces langues de l'anglais et du français", 
    "Modèl espesyalize sa a depase zouti ki egziste deja pou kreyòl yo."]

labels = model.predict(texts)
for i, text in enumerate(texts):
    print(labels[i], text, sep=": ")
```
